# Wind power and capacity factors

Code-only usage example adapted from step4_result_chek.ipynb. Set an external input path. These are simulated outputs; total_generation is power in kW, not measured energy. Run with this notebook directory as the working directory.

```python
from pathlib import Path
import xarray as xr
from analyze import hourly_frame, daily_frame

INPUT = Path("/path/to/wind_aggregated_by_province.nc")
PROVINCE = "云南省"
```

```python
with xr.open_dataset(INPUT) as ds:
    print(ds)
    hourly = hourly_frame(ds, PROVINCE)
daily = daily_frame(hourly)
hourly.head()
```

## Daily energy

For regular hourly power, sum(kW × 1 hour) / 1000 gives MWh. Dates follow source timestamps without an assumed timezone shift. Incomplete boundary days are flagged and omitted below.

```python
daily.loc[daily.complete_day, "energy_mwh"].plot(ylabel="Simulated energy (MWh/day)")
```

## Monthly capacity factor

Arithmetic mean across grids and capacity-weighted CF answer different questions; retain both.

```python
hourly[["mean_capacity_factor", "weight_capacity_factor"]].resample("MS").mean().plot(ylabel="Capacity factor (1)")
```
